```python
import pandas as pd

# Load the dataset
df = pd.read_csv('StudentsPerformance.csv')  # Make sure the file is in your Colab or local folder

# Preview the data
df.head()
```

```python
# Basic info about the dataset
df.info()

# Summary statistics
df.describe()

# Check column names
df.columns
```

```python
# Check for missing values
df.isnull().sum()
```

```python
# Mean scores by gender
df.groupby('gender')[['math score', 'reading score', 'writing score']].mean()

# Mean scores by lunch type
df.groupby('lunch')[['math score', 'reading score', 'writing score']].mean()

# Mean scores by test preparation
df.groupby('test preparation course')[['math score', 'reading score', 'writing score']].mean()

# Mean scores by parental education
df.groupby('parental level of education')[['math score', 'reading score', 'writing score']].mean()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

# Boxplot: Math scores by gender
sns.boxplot(x='gender', y='math score', data=df)
plt.title('Math Score Distribution by Gender')
plt.show()

# Histogram: Reading score
sns.histplot(df['reading score'], kde=True)
plt.title('Reading Score Distribution')
plt.show()

# Barplot: Average scores by lunch type
avg_scores = df.groupby('lunch')[['math score', 'reading score', 'writing score']].mean().reset_index()
avg_scores.plot(x='lunch', kind='bar')
plt.title('Average Scores by Lunch Type')
plt.xticks(rotation=0)
plt.show()
```
